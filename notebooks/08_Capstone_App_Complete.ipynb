## 1.1. Organisation du code

Au fil des notebooks, nous avons progressivement structuré le projet pour séparer les responsabilités. Prenons un moment pour faire le point sur cette architecture, car c'est elle qui rend votre application maintenable dans la durée.

Le fichier [utils/data.py](../streamlit_app/utils/data.py) centralise tout ce qui touche au chargement et au traitement des données. Si demain vous changez de source (une base de données au lieu d'un CSV, par exemple), vous n'aurez qu'un seul fichier à modifier. Le fichier [utils/charts.py](../streamlit_app/utils/charts.py) joue le même rôle pour les graphiques Plotly : vos fonctions de visualisation sont réutilisables d'une page à l'autre sans dupliquer le code.

Le dossier [pages/](../streamlit_app/pages/) contient les différentes vues de votre application. C'est le mécanisme **multipage** natif de Streamlit : chaque fichier Python dans ce dossier devient automatiquement une page accessible dans la barre de navigation latérale. Le dossier [data/](../streamlit_app/data/) regroupe vos jeux de données, et [.streamlit/](../streamlit_app/.streamlit/) contient la configuration du thème.

💡 Cette organisation n'est pas propre à Streamlit — vous la retrouverez dans la plupart des projets Python bien structurés. L'idée centrale est toujours la même : un fichier = une responsabilité.

## 1.2. Les briques Streamlit que vous maîtrisez

Avant de passer au packaging, faisons un rapide tour d'horizon des principales méthodes Streamlit que vous avez utilisées au cours de cette formation. Vous les retrouverez dans votre code, et c'est utile de les avoir en tête pour la relecture.

`st.set_page_config` vous a permis de configurer le titre et le layout de votre application dès la première ligne. `st.cache_data` a résolu le problème de performance en mettant en cache le résultat de vos chargements de données — sans cette méthode, Streamlit rechargerait le CSV à chaque interaction utilisateur.

Pour construire l'interface, vous avez utilisé les widgets de saisie comme `st.selectbox`, `st.slider` et `st.date_input`, dont les valeurs de retour alimentent directement vos filtres. Le `st.session_state` vous a permis de conserver des informations entre deux interactions, et `st.query_params` d'écrire l'état des filtres dans l'URL pour créer des permaliens.

Côté mise en page, `st.columns` et `st.tabs` vous ont aidé à organiser le contenu, tandis que `st.download_button` a rendu possible l'export de fichiers (CSV, PNG, ZIP) en un clic. Et bien sûr, `st.plotly_chart` a fait le lien entre vos figures Plotly et l'interface Streamlit.

📖 La [référence API de Streamlit](https://docs.streamlit.io/develop/api-reference) est le meilleur endroit pour retrouver le détail de chaque méthode quand vous en avez besoin.

## 1.5. Checklist qualité

Avant de partager votre application, prenez quelques minutes pour vérifier ces points essentiels.

Commencez par les **graphiques** : chaque visualisation doit avoir un titre explicite et des axes avec des unités visibles. Les légendes doivent être lisibles sans avoir besoin de zoomer. Passez ensuite aux **filtres** : testez-les tous, y compris les cas limites (aucune donnée sélectionnée, une seule catégorie, plage de dates vide). Vérifiez que les **exports** (PNG, CSV, ZIP) fonctionnent correctement — téléchargez-les et ouvrez-les pour vous en assurer. Enfin, contrôlez la **performance** : si votre application met plus de quelques secondes à charger, vérifiez que vous utilisez bien `st.cache_data` sur vos fonctions de chargement de données.

💡 Cette checklist paraît simple, mais elle couvre la majorité des problèmes que vos utilisateurs rencontreront. Mieux vaut passer cinq minutes à tout vérifier que de recevoir un message « ça ne marche pas » une heure après le partage.
