# 1. Créer l'arborescence minimale du projet

Avant d'écrire la moindre ligne de code Streamlit, nous allons structurer notre projet. C'est une habitude essentielle en développement : un projet bien organisé dès le départ, c'est un projet qui sera facile à faire évoluer et à maintenir.

Voici l'arborescence que nous allons mettre en place :

```
streamlit_app/
  app.py
  pages/
  utils/
  data/
  .streamlit/config.toml
```

Chaque élément a un rôle bien précis, que nous allons détailler dans les sections suivantes.

Commençons par comprendre pourquoi cette structure est importante, avant de la créer concrètement avec du code Python.

## 1.1. Pourquoi structurer son projet ?

Quand vous travaillez sur un notebook Jupyter, tout tient dans un seul fichier : c'est pratique pour l'exploration, mais dès que vous passez à une application web, les choses changent. Vous allez avoir du code d'interface, des fonctions utilitaires, des données, de la configuration…

Séparer tout cela dans des dossiers dédiés a plusieurs avantages concrets. D'abord, la **lisibilité** : même quelques mois plus tard, vous saurez immédiatement où trouver ce que vous cherchez. Ensuite, la **réutilisabilité** : vos fonctions utilitaires pourront être reprises dans d'autres projets. La **collaboration** est aussi facilitée, car vos collègues retrouveront une structure qu'ils connaissent. Enfin, l'**évolutivité** : ajouter une page ou une source de données restera simple et sans risque.

## 1.2. L'arborescence que nous utiliserons

Voici le détail de chaque élément de notre arborescence :

```
streamlit_app/
    app.py                  ← Point d'entrée de l'application
    pages/                  ← Pages additionnelles (menu automatique)
    utils/                  ← Fonctions utilitaires réutilisables
    data/                   ← Jeux de données (CSV, etc.)
    .streamlit/config.toml  ← Configuration du thème
```

Le fichier `app.py` est le cœur de votre application : c'est le premier fichier que Streamlit va exécuter. Le dossier `pages/` est particulièrement intéressant car Streamlit détecte automatiquement les fichiers `.py` placés dedans et les ajoute comme pages dans le menu de navigation — nous y reviendrons plus tard. Le dossier `utils/` accueillera vos fonctions de nettoyage, de transformation ou de chargement de données. Enfin, `data/` stockera vos fichiers de données, et `.streamlit/config.toml` contiendra la configuration visuelle de votre application.

## 1.3. Création automatique des dossiers avec Python

Plutôt que de créer ces dossiers manuellement, nous allons utiliser le module [`pathlib`](https://docs.python.org/3/library/pathlib.html) de Python. Ce module permet de manipuler les chemins de fichiers de manière moderne et fonctionne de la même façon sous Windows, Mac et Linux.

La cellule suivante crée toute l'arborescence en une seule exécution :

````python
from pathlib import Path

base = Path("../streamlit_app")
for p in [base, base/"pages", base/"utils", base/"data", base/".streamlit"]:
    p.mkdir(parents=True, exist_ok=True)

print("Arborescence créée :", base.resolve())
````

## 1.4. Bonnes pratiques à retenir

💡 Quelques réflexes à adopter dès le départ : séparez toujours le code, les données et la configuration dans des dossiers distincts. Utilisez des noms de fichiers et de dossiers explicites — un collègue doit pouvoir comprendre votre projet rien qu'en lisant l'arborescence. Pensez aussi à ajouter un fichier `README.md` à la racine pour documenter la structure et les instructions de lancement.

En suivant ces habitudes, vous posez des bases solides pour un projet professionnel, facile à maintenir et à faire évoluer.

# 2. Un premier `app.py` minimal

Maintenant que notre arborescence est en place, nous allons écrire le fichier `app.py` — le point d'entrée de votre application Streamlit.

Concrètement, Streamlit fonctionne de manière assez inhabituelle pour du développement web : vous écrivez un **script Python classique** (pas de HTML, pas de JavaScript), et Streamlit se charge de le transformer en page web interactive. Chaque fois que vous appelez une fonction comme `st.title()` ou `st.write()`, Streamlit ajoute un élément à la page, de haut en bas, dans l'ordre de votre script.

Voyons les étapes une par une.

## 2.1. Importer les librairies nécessaires

Nous commençons par importer deux librairies :

```python
import streamlit as st
import plotly.express as px
```

[`streamlit`](https://docs.streamlit.io/library/get-started) est le framework qui va transformer votre script Python en application web. Par convention, nous l'importons sous l'alias `st` — vous verrez que toutes les fonctions Streamlit s'appellent avec ce préfixe (par exemple `st.title()`, `st.write()`, etc.).

[`plotly.express`](https://plotly.com/python/plotly-express/) est le module "haut niveau" de Plotly, qui permet de créer des graphiques interactifs en une seule ligne de code. Nous l'importons sous l'alias `px`.

## 2.2. Configurer la page Streamlit

La toute première chose à faire dans votre `app.py` est d'appeler [`st.set_page_config()`](https://docs.streamlit.io/library/api-reference/utilities/st.set_page_config). Cette méthode configure les paramètres "globaux" de votre page : le titre qui apparaît dans l'onglet du navigateur, l'icône de l'onglet, et la disposition de la page.

```python
st.set_page_config(page_title="DataViz — Démo", page_icon="📊", layout="wide")
```

Le paramètre `layout="wide"` est important : il indique à Streamlit d'utiliser toute la largeur de l'écran plutôt qu'une colonne centrale étroite. Pour un dashboard de data visualisation, c'est souvent préférable car vos graphiques auront plus de place pour s'afficher.

⚠️ `st.set_page_config()` doit **toujours** être le premier appel Streamlit de votre script, avant tout autre `st.xxx()`. Sinon, vous obtiendrez une erreur.

## 2.3. Ajouter un titre principal

La méthode [`st.title()`](https://docs.streamlit.io/library/api-reference/text/st.title) affiche un titre de niveau 1 en haut de votre page. C'est l'équivalent d'une balise `<h1>` en HTML, mais vous n'avez pas besoin de connaître le HTML — Streamlit s'en occupe pour vous.

```python
st.title("📊 Démo Streamlit + Plotly")
```

Streamlit propose d'autres méthodes pour afficher du texte avec différents niveaux de hiérarchie : `st.header()` pour un titre de niveau 2, `st.subheader()` pour un niveau 3, et `st.markdown()` pour du texte formaté en Markdown. Nous les utiliserons au fil des prochains notebooks.

## 2.4. Afficher du texte

La méthode [`st.write()`](https://docs.streamlit.io/library/api-reference/write-magic/st.write) est la méthode "couteau suisse" de Streamlit. Elle détecte automatiquement le type de ce que vous lui passez et l'affiche de manière appropriée : du texte brut, du Markdown, un DataFrame Pandas, un graphique Plotly… Dans notre cas, nous l'utilisons simplement pour afficher un message de bienvenue.

```python
st.write("Bienvenue ! Ceci est votre toute première page d'application.")
```

💡 `st.write()` est très pratique pour prototyper rapidement. Pour un contrôle plus fin du formatage, vous pourrez utiliser `st.markdown()` qui accepte la syntaxe Markdown complète.

## 2.5. Charger un jeu de données d'exemple

Pour tester nos visualisations sans avoir à chercher un fichier CSV, nous allons utiliser un jeu de données intégré directement dans Plotly. La fonction [`px.data.iris()`](https://plotly.com/python-api-reference/generated/plotly.express.data.iris.html) retourne un DataFrame Pandas contenant le célèbre dataset Iris (mesures de sépales et pétales pour trois espèces de fleurs).

```python
df = px.data.iris()
```

Dans les prochains notebooks, nous remplacerons ce jeu de données par vos propres fichiers CSV. Pour l'instant, il nous suffit d'avoir des données à afficher.

## 2.6. Créer un graphique interactif

La fonction [`px.scatter()`](https://plotly.com/python-api-reference/generated/plotly.express.scatter.html) de Plotly Express permet de créer un nuage de points (*scatter plot*) interactif en une seule ligne. Vous lui passez votre DataFrame et vous précisez quelles colonnes utiliser pour les axes X, Y et la couleur.

```python
fig = px.scatter(
    df, x="sepal_width", y="sepal_length", color="species",
    title="Iris — largeur vs longueur des sépales"
)
```

Le résultat est un objet `fig` (une figure Plotly) que vous pouvez ensuite afficher. Les graphiques Plotly sont **nativement interactifs** : vos utilisateurs pourront zoomer, survoler les points pour voir les valeurs, et même masquer des catégories en cliquant sur la légende.

## 2.7. Afficher le graphique dans Streamlit

Pour intégrer un graphique Plotly dans votre application Streamlit, vous utilisez la méthode [`st.plotly_chart()`](https://docs.streamlit.io/library/api-reference/charts/st.plotly_chart). Elle prend en paramètre la figure Plotly et l'affiche directement dans la page.

```python
st.plotly_chart(fig, use_container_width=True)
```

Le paramètre `use_container_width=True` fait en sorte que le graphique occupe toute la largeur disponible dans la page, ce qui donne un rendu plus professionnel, surtout en `layout="wide"`.

Voilà ! Avec ces quelques lignes, vous avez une application Streamlit fonctionnelle qui affiche un graphique interactif. La cellule suivante va écrire tout ce code dans le fichier `app.py`.

````python
%%writefile ../streamlit_app/app.py
import streamlit as st  # Importation de Streamlit pour créer l'interface web
import plotly.express as px  # Importation de Plotly Express pour les graphiques interactifs

# Configuration de la page Streamlit (titre, icône, mise en page)
st.set_page_config(page_title="DataViz — Démo", page_icon="📊", layout="wide")

# Titre principal de l'application
st.title("📊 Démo Streamlit + Plotly")
# Texte d'introduction
st.write("Bienvenue ! Ceci est votre toute première page d’application.")

# Chargement d'un jeu de données exemple intégré à Plotly
df = px.data.iris()

# Création d'un graphique en nuage de points (scatter plot)
fig = px.scatter(
    df, x="sepal_width", y="sepal_length", color="species",
    title="Iris — largeur vs longueur des sépales"
)

# Affichage du graphique Plotly dans l'application Streamlit
st.plotly_chart(fig, use_container_width=True)
````

# 3. Configurer un thème visuel

Par défaut, Streamlit applique un thème générique à votre application. Nous allons le personnaliser dès maintenant via un fichier de configuration [`config.toml`](https://docs.streamlit.io/develop/api-reference/configuration/config.toml), placé dans le dossier `.streamlit/`.

Un fichier **TOML** (*Tom's Obvious, Minimal Language*) est un format de configuration simple et lisible — vous pouvez le voir comme un fichier texte structuré en sections, avec des paires clé/valeur. Dans notre cas, nous allons définir trois paramètres : le thème de base (clair ou sombre), la couleur principale de l'interface, et la police d'écriture.

## 3.1. Pourquoi personnaliser le thème dès le départ ?

Même si cela peut sembler secondaire à ce stade, définir un thème visuel dès le début de votre projet est une bonne habitude. Cela garantit une **cohérence graphique** sur toutes les pages de votre application : les boutons, les liens et les éléments interactifs utiliseront tous la même couleur principale. Cela renforce aussi la crédibilité de votre dashboard auprès de vos utilisateurs, et vous pourrez facilement l'adapter aux couleurs de votre entreprise le moment venu.

## 3.2. Les options de thème disponibles

Streamlit expose plusieurs paramètres de personnalisation dans la section `[theme]` du fichier `config.toml`. Voici les trois que nous allons utiliser :

- **`base`** — Le thème de base, soit `"light"` (fond blanc) soit `"dark"` (fond sombre). C'est le point de départ de toute la palette de couleurs.
- **`primaryColor`** — La couleur principale de l'interface, appliquée aux boutons, liens, sliders et éléments interactifs. Nous utiliserons `"#0F766E"`, un vert-bleu professionnel.
- **`font`** — La police d'écriture. Streamlit propose `"sans serif"`, `"serif"` ou `"monospace"`.

## 3.3. Mettre en place le fichier de configuration

La mise en place est simple. Le fichier `config.toml` doit se trouver dans le dossier `.streamlit/` à la racine de votre projet (nous l'avons déjà créé dans l'arborescence à la section 1). Voici à quoi il ressemble :

```toml
[theme]
base="light"
primaryColor="#0F766E"
font="sans serif"
```

La cellule de code qui suit va écrire ce contenu dans le fichier. Après toute modification de `config.toml`, il vous suffira de relancer votre application Streamlit pour voir le nouveau thème appliqué.

## 3.4. Pour aller plus loin

📖 La documentation officielle de Streamlit détaille [toutes les options de thème disponibles](https://docs.streamlit.io/develop/api-reference/configuration/config.toml#theme-options), y compris les couleurs d'arrière-plan, les couleurs de texte, et bien d'autres.

💡 Définir un thème dès le départ est une excellente habitude. Même si vous le modifierez probablement au fil du projet, avoir cette base en place vous évitera de devoir refaire toute la charte graphique plus tard.

````python
%%writefile ../streamlit_app/.streamlit/config.toml
# Définition du thème visuel Streamlit via le fichier de configuration TOML

[theme]  # Section de configuration du thème
base="light"  # Thème clair par défaut
primaryColor="#0F766E"  # Couleur principale (boutons, liens, etc.)
font="sans serif"  # Police d’écriture utilisée dans l’interface
````
